```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create sample dataset
data = {
    "Area_sqft": [500, 700, 900, 1100, 1300, 1500],
    "Price": [100000, 140000, 180000, 220000, 260000, 300000]
}

df = pd.DataFrame(data)

# Feature (input)
X = df[["Area_sqft"]]

# Label (output)
y = df["Price"]

# Train Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Predict house price for a new house
new_area = pd.DataFrame({"Area_sqft": [1200]})
predicted_price = model.predict(new_area)

print("Predicted price for a 1200 sqft house:", predicted_price[0])
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Area_sqft"], df["Price"])
plt.plot(df["Area_sqft"], model.predict(X))
plt.xlabel("Area (sqft)")
plt.ylabel("House Price")
plt.title("House Price Prediction using Linear Regression")
plt.show()
```
