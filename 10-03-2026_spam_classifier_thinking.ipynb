## 1. Features for Spam Detection
Features are the characteristics used by the model to identify spam messages.

Possible features:
- Presence of suspicious words (e.g., "free", "win", "offer", "urgent")
- Number of links in the message
- Use of excessive capital letters
- Presence of special characters (!, $, %)
- Length of the message
- Sender email/domain reputation

## 2. Data Needed
To train a spam classifier, we need a labeled dataset containing:

- Emails or messages
- Label indicating whether each message is **Spam** or **Not Spam**

Example dataset format:

| Message | Label |
|-------|-------|
| "Win a free iPhone now!" | Spam |
| "Meeting scheduled at 3 PM" | Not Spam |
| "Limited time offer click here" | Spam |

Large datasets with thousands of examples help the model learn better patterns.

## 3. Possible Mistakes (Model Errors)

### False Positive
A normal email is incorrectly classified as spam.

Example:
"Project update meeting tomorrow" → marked as Spam.

### False Negative
A spam message is incorrectly classified as normal.

Example:
"You won a lottery prize!" → marked as Not Spam.

Both types of mistakes affect user experience and system reliability.
