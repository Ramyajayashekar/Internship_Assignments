### Resume Keyword Matcher using NLP

Recruiters manually review resumes, which is time-consuming and inconsistent. This project aims to automate resume screening by comparing resumes with job descriptions and generating a match score along with missing skills.

### Type of Data

Text-based dataset consisting of:

Resume content
Job descriptions
🔹 Example Dataset
Resume Text	Job Description
"Python, Machine Learning, Data Analysis"	"Looking for Python and ML skills"
"Java, SQL, Web Development"	"Need backend developer with Java"
🔹 How Dataset is Collected
Sample resumes (manually created or from open sources)
Job descriptions from platforms like LinkedIn or Naukri
🔹 Preprocessing Steps
Convert text to lowercase
Remove punctuation
Remove stopwords
Tokenization

### Algorithm

### Step 1: Input
Take Resume Text (R)
Take Job Description (J)
### Step 2: Text Preprocessing
Convert both R and J to lowercase
Remove punctuation and special characters
Tokenize text into words
Remove stopwords (e.g., “the”, “is”, “and”)
### Step 3: Feature Extraction
Apply TF-IDF on both texts
Convert R and J into numerical vectors

 Output:

Vector_R
Vector_J
### Step 4: Similarity Calculation
Compute similarity using Cosine Similarity
Formula:
Similarity=
∣∣R∣∣×∣∣J∣∣
R⋅J
	​

### Step 5: Match Score
Convert similarity into percentage
Match Score=Similarity×100
### Step 6: Missing Skills Detection
Extract keywords from job description
Compare with resume words
Identify missing terms
### Step 7: Output
Display:
Match Score (%)
Missing Skills
Matched Skills
