```python
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# Slang dictionary
slang_dict = {
    "omg": "oh my god",
    "lol": "laughing",
    "idk": "i do not know",
    "brb": "be right back",
    "u": "you",
    "r": "are",
    "ur": "your",
    "thx": "thanks",
    "pls": "please",
    "gr8": "great",
    "luv": "love",
    "rn": "right now",
    "w8": "wait",
    "nt": "not"
}

def expand_slang(text):
    words = text.split()
    return " ".join([slang_dict.get(word, word) for word in words])

def remove_emojis(text):
    return re.sub(r'[^\x00-\x7F]+', '', text)

def clean_text(text):
    text = text.lower()
    text = expand_slang(text)
    text = remove_emojis(text)
    
    # Remove punctuation & numbers
    text = re.sub(r'[^a-z\s]', '', text)
    
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    # Remove stopwords
    words = text.split()
    words = [w for w in words if w not in ENGLISH_STOP_WORDS]
    
    return " ".join(words)

# 🔥 More test examples
sentences = [
    "OMG!!! I got this for FREE 😍🔥",
    "plsssss send the file asap!!!",
    "LOL 😂 that was EPIC!!!",
    "u r amazing bro!!! 💯",
    "idk why this isnt working???",
    "this movie is soooo boring 😴😴",
    "BRB, will call u in 10 mins",
    "thx a lot for ur support 🙏",
    "gimme the details rn!!!",
    "cant w8 to see this!!! 🤩",
    "this is nt good at all!!!",
    "hahahaha that meme 😂😂",
    "meet me @ cafe @ 6pm",
    "i luv this product!!! ❤️",
    "wat is going on here???",
    "soooo excited for the trip!!! ✈️",
    "pls fix this issue asap!!!",
    "u gotta try this food 😋",
    "this is awsm bro!!!",
    "omg omg omg this is crazy!!!"
]

# Display output
for i, s in enumerate(sentences, 1):
    print(f"{i}. Original : {s}")
    print(f"   Cleaned  : {clean_text(s)}\n")
```
