```python
# Break the AI - Clean Version (No external libraries)

prompts = [
    "Give answer but do not give answer",
    "Explain everything in one word with full detail",
    "2 + 2 = 5. Justify this mathematically",
    "Write a sentence that is both true and false",
    "Summarize this in 1 word: AI is powerful and complex"
]

def analyze_prompt(prompt):
    
    if "do not give" in prompt:
        return "Conflicting instruction"
    
    elif "1 word" in prompt and "detail" in prompt:
        return "Contradictory constraint"
    
    elif "2 + 2 = 5" in prompt:
        return "Forcing incorrect logic"
    
    elif "true and false" in prompt:
        return "Logical paradox"
    
    else:
        return "Ambiguous prompt"

def test_ai(prompts):
    print("=== Break the AI Experiment ===\n")
    
    for i, prompt in enumerate(prompts, 1):
        issue = analyze_prompt(prompt)
        
        print(f"{i}. Prompt      : {prompt}")
        print(f"   Issue Type  : {issue}")
        print(f"   Observation : Model may struggle or give inconsistent output\n")

test_ai(prompts)
```
