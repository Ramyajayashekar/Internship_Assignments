```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
print(df)
```

```python
df['maths'] = df['maths'].fillna(df['maths'].mean())
df['science'] = df['science'].fillna(df['science'].mean())
df['english'] = df['english'].fillna(df['english'].mean())
```

```python
df['total'] = df['maths'] + df['science'] + df['english']
```
