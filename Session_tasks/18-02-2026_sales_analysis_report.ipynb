```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Month': ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    'Revenue': [20000,22000,25000,27000,30000,32000,35000,37000,36000,39000,42000,45000],
    'Marketing_Spend': [3000,3200,3500,4000,4200,4500,4700,5000,4800,5200,5500,6000],
    'Profit': [8000,9000,10000,11000,12500,13500,15000,16000,15500,17000,18500,20000]
}

df = pd.DataFrame(data)
print(df)
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['Month'], df['Revenue'], marker='o')
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df['Marketing_Spend'], df['Profit'])
plt.title("Marketing Spend vs Profit")
plt.xlabel("Marketing Spend")
plt.ylabel("Profit")
plt.show()
```

```python
corr = df[['Revenue','Marketing_Spend','Profit']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
